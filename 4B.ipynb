```python

# Assignment 4B: CBOW Model for Twitter Dataset

# Install if required
# %pip install gensim pandas scikit-learn

# 1. Import the required libraries
import pandas as pd
import re

from gensim.models import Word2Vec
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS


# 2. Load the Twitter dataset
df = pd.read_csv("twitter_dataset.csv")


# 3. Preprocess the tweets
stop_words = set(ENGLISH_STOP_WORDS)

def preprocess(text):
    # Convert text to lowercase
    text = str(text).lower()

    # Remove URLs and Twitter usernames
    text = re.sub(r"http\S+|www\S+|@\w+", " ", text)

    # Keep only lowercase letters and whitespace
    text = re.sub(r"[^a-z\s]", " ", text)

    # Split the text into individual words
    words = text.split()

    # Remove stop words
    clean_words = []
    for word in words:
        if word not in stop_words:
            clean_words.append(word)

    return clean_words


# 4. Tokenize and clean all tweets
tweets = []

for text in df["Text"].dropna():
    cleaned = preprocess(text)

    if len(cleaned) > 0:
        tweets.append(cleaned)

print("\nTotal Tweets:", len(tweets))

print("Sample Processed Tweet:", tweets[0])


# 5. Train the Word2Vec model using CBOW
model = Word2Vec(
    sentences=tweets,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=20
)


# 6. Inspect a learned word embedding

word = model.wv.index_to_key[0]

print("\nSelected Word:", word)
print("Word Vector:\n", model.wv[word])


# 7. Find the most similar words
print("\nMost Similar Words:")
print(model.wv.most_similar(word, topn=5))


# 8. Calculate similarity between two words
word1 = model.wv.index_to_key[0]
word2 = model.wv.index_to_key[1]

similarity = model.wv.similarity(word1, word2)

print(
    "\nSimilarity between",
    word1,
    "and",
    word2,
    ":",
    similarity
)

```
