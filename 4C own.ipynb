```python
with open (r"document3.txt", "r", encoding='utf-8') as file:
    text = file.read()
```

```python
text
```

```python
text = text.lower()

sentences = re.split(r"[.!?]+", text)

data = []

for sen in sentences:
    words = re.findall(r"[a-z]+", sen)

    if len(words) > 0:
        data.append(words)
```

```python
model = Word2Vec(
    sentences=data,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    epochs=100
)
```

```python
word = "influenza"

print("\nSelected word: ", word)
print("\nWord vector: ", model.wv[word])
```

```python
print("\nMost similar words to ", word, "\n")
print(model.wv.most_similar(word, topn=3))
```

```python
word1 = "covid"
word2 = "influenza"

print("\nSimilarity between covid and influenza: ", model.wv.similarity(word1, word2))
```
