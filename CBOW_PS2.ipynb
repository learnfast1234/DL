```python
import pandas as pd
import re
from gensim.models import Word2Vec
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
```

```python
 df = pd.read_csv(r"D:\ML\DL\LP IV Updated Datasets\CBOW\twitter_dataset.csv")
```

```python
df.head()
```

```python
stop_words = set(ENGLISH_STOP_WORDS)
```

```python
def preprocess(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+|@w+'," ",text)

    text = re.sub(r"[^a-z\s]"," ",text)

    words = text.split()

    clean_words = []

    for word in words:
        if word not in stop_words:
            clean_words.append(word)
    return clean_words
```

```python
tweets = []
for text in df["Text"].dropna():
    cleaned = preprocess(text)

    if len(cleaned) > 0:
        tweets.append(cleaned)
print("\nTotal Tweets:", len(tweets))
print("Sample Proceseed Tweet:",tweets[0])
```

```python
model = Word2Vec(
    sentences = tweets,
    vector_size = 100,
    window=5,
    min_count = 2,
    epochs = 20
)
    
```

```python
print("\nMost Similar Words:")
print(model.wv.most_similar(word, topn=5))
```

```python
word1 = model.wv.index_to_key[0]
word2 = model.wv.index_to_key[1]
similarity = model.wv.similarity(word1, word2)
print("\nSimilarity between", word1, "and", word2, ":", similarity)
```
