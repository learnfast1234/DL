```python
df = pd.read_csv('twitter_dataset.csv')

df.head()
```

```python
#preprocess

stop_words = set(ENGLISH_STOP_WORDS)

def preprocess(text):

    text = str(text).lower()

    text = re.sub(r"http\S+|www\S+|@\w+", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)

    words = text.split()

    filtered_words = []

    for word in words:
        if word not in stop_words:
            filtered_words.append(word)


    return filtered_words

data = []
for text in df['Text'].dropna():
    cleaned = preprocess(text)

    if len(cleaned) > 0:
        data.append(cleaned)
            
```

```python
#now I have the tokenized words

model = Word2Vec(
    sentences=data,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=20
)
```

```python
word = model.wv.index_to_key[0]

print("\nSelected word: ", word)
print("\nWord vector: ", model.wv[word])
```

```python
print("\nMost similar words to ", word, "\n")
print(model.wv.most_similar(word, topn=5))
```

```python
word1 = model.wv.index_to_key[0]
word2 = model.wv.index_to_key[1]

print("Similarity between ", word1, " and ", word2, ": ", model.wv.similarity(word1, word2))
```
