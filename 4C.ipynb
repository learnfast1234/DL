```python

# Assignment 4C: CBOW Model for Textual Document

# Install if required
# %pip install gensim

# 1. Import the required libraries
import re
from gensim.models import Word2Vec


# 2. Load the textual document
with open(
    r"D:\ML\DL\LP IV Updated Datasets\CBOW\document3.txt",
    "r",
    encoding="utf-8"
) as file:
    text = file.read()


# 3. Convert the text to lowercase
text = text.lower()


# 4. Split the document into sentences using punctuation
sentences = re.split(r'[.!?]+', text)


# 5. Tokenize each sentence into words
data = []

for sen in sentences:
    # Extract lowercase English words
    words = re.findall(r'[a-z]+', sen)

    # Add non-empty word lists to the training data
    if len(words) > 0:
        data.append(words)


# 6. Train the Word2Vec model using CBOW
model = Word2Vec(
    sentences=data,
    vector_size=50,
    window=3,
    sg=0,           # sg=0 selects CBOW
    min_count=1,    # Keep words occurring at least once
    epochs=100
)


# 7. Display the learned vocabulary
print("Vocabulary:", model.wv.index_to_key)

# Display the number of unique words in the vocabulary
print("Vocabulary Size:", len(model.wv))


# 8. Display the embedding of a selected word
word = "influenza"
print("\nWord:", word)
print("Word Vector:", model.wv[word])


# 9. Find the three most similar words to "influenza"
print("\nMost Similar Words:")
print(model.wv.most_similar("influenza", topn=3))


# 10. Calculate similarity between "covid" and "influenza"
print(
    "\nSimilarity between covid and influenza:",
    model.wv.similarity("covid", "influenza")
)
```
