```python
import re
from gensim.models import Word2Vec
```

```python
with open("D:\ML\DL\LP IV Updated Datasets\CBOW\document3.txt","r",encoding="utf-8") as file:
    text = file.read()
```

```python
text
```

```python
text = text.lower()
text 
```

```python
sentences = re.split(r'[.!?]+',text)
```

```python
data = []
for sen in sentences :
    words = re.findall(r'[a-z]+',sen)
    if len(words) > 0:
        data.append(words)
    
```

```python
data
```

```python
model = Word2Vec(
    sentences= data,
    vector_size = 50,
    window = 3,
    sg = 0,
    min_count = 1,
    epochs=100
    
)
```

```python
print("Vocabulary : ",model.wv.index_to_key)
```

```python
len(model.wv)
```

```python
word = "influenza"
model.wv[word]
```

```python
model.wv.most_similar("influenza",topn=3)
```
